# Health Data Start Here

Use this notebook after importing your Fitbit export. It connects to the health database through `HEALTH_DATABASE_URL`.

```python
import os
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(os.environ["HEALTH_DATABASE_URL"])
pd.read_sql("select * from import_runs order by started_at desc limit 10", engine)
```

```python
heart_rate = pd.read_sql("""
select ts, value as bpm
from metric_samples
where metric_type = 'heart_rate'
order by ts
""", engine, parse_dates=["ts"])

heart_rate.set_index("ts").resample("1h").mean().plot(figsize=(14, 4), title="Hourly average heart rate")
```

```python
sleep = pd.read_sql("""
select date_of_sleep, duration_seconds / 3600.0 as hours_asleep, efficiency
from sleep_sessions
where date_of_sleep is not null
order by date_of_sleep
""", engine, parse_dates=["date_of_sleep"])

sleep.set_index("date_of_sleep")[["hours_asleep", "efficiency"]].tail(60)
```
